```python
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-5, 5, 500)

# Activation Functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def elu(x, alpha=1.0):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

def softmax(x):
    exp_x = np.exp(x - np.max(x))   # Numerical stability
    return exp_x / np.sum(exp_x)

# Compute outputs
y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)
y_leaky = leaky_relu(x)
y_elu = elu(x)
y_softmax = softmax(x)

# Plotting
plt.figure(figsize=(14, 10))

# Sigmoid
plt.subplot(2, 3, 1)
plt.plot(x, y_sigmoid, color='blue')
plt.title("Sigmoid")
plt.grid(True)

# Tanh
plt.subplot(2, 3, 2)
plt.plot(x, y_tanh, color='green')
plt.title("Tanh")
plt.grid(True)

# ReLU
plt.subplot(2, 3, 3)
plt.plot(x, y_relu, color='red')
plt.title("ReLU")
plt.grid(True)

# Leaky ReLU
plt.subplot(2, 3, 4)
plt.plot(x, y_leaky, color='purple')
plt.title("Leaky ReLU")
plt.grid(True)

# ELU
plt.subplot(2, 3, 5)
plt.plot(x, y_elu, color='orange')
plt.title("ELU")
plt.grid(True)

# Softmax
plt.subplot(2, 3, 6)
plt.plot(x, y_softmax, color='brown')
plt.title("Softmax")
plt.grid(True)

plt.tight_layout()
plt.show()
```
